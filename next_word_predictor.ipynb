```python
import numpy as np
import pandas as pd
import tensorflow
from tensorflow import keras
from tensorflow.keras.layers import *
from tensorflow.keras.models import Sequential
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import to_categorical
```

```python
with open("data.txt", "r", encoding="utf-8") as file:
    text = file.read()

print(text)
```

```python
tokenizer=Tokenizer() 
tokenizer.fit_on_texts([text])
```

```python
tokenizer.word_index
```

```python
len(tokenizer.word_index)
```

```python
data = [s.strip() for s in text.split('.') if s.strip()]
text_data_final=[] 
for sentence in data: 
    text_data=tokenizer.texts_to_sequences([sentence]) 
    text_data_final.append(text_data[0]) 
print(text_data_final)
```

```python
text_data_input=[]
text_data_output=[]
for data in text_data_final:
    for i in range(1,len(data)):
        text_data_input.append(data[0:i])
        text_data_output.append(data[i])
text_data_output = np.array(text_data_output)
```

```python
text_data_output_final = to_categorical(text_data_output,num_classes=4994)
```

```python
max_len=0
for data in text_data_input:
    if len(data)>max_len:
        max_len=len(data)
print(max_len)
```

```python
text_data_final_padded = pad_sequences(text_data_input,maxlen=max_len,padding='pre')
```

```python
text_data_final_padded
```

```python
model = Sequential() 
model.add(Input(shape=(83,)))
model.add(Embedding(4994,64))
model.add(LSTM(150,return_sequences=True))
model.add(LSTM(100))
model.add(Dense(4994,activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(loss='categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
history=model.fit(text_data_final_padded,text_data_output_final,epochs=50)
```

```python
text = "I love"
for i in range(5):
    output=tokenizer.texts_to_sequences([text])
    output_padded=pad_sequences(output,maxlen=max_len,padding='pre')
    predicted_word_index=np.argmax(model.predict(output_padded)[0])
    predicted_word = tokenizer.index_word[predicted_word_index]
    print(text + ' ' + predicted_word)
    text = text + ' ' + predicted_word
```
